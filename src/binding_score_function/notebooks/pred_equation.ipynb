# Regression fit

```python
train_regression_predictions = pd.read_csv("/home/er8813ha/immunopeptides/plots_v2_both/regression/train_predictions.csv")
val_regression_predictions = pd.read_csv("/home/er8813ha/immunopeptides/plots_v2_both/regression/val_predictions.csv")
regression_scaling_parameters = pd.read_csv("/home/er8813ha/immunopeptides/plots_v2_both/regression/scaling_params.csv")
train_regression_predictions.head()
```

```python
symbolic_equations = pd.read_csv("/home/er8813ha/immunopeptides/plots_v2_both/regression/symbolic_regression_all_equations.csv")
symbolic_equations
```

```python
plt.figure(figsize=(4,4))
sns.scatterplot(symbolic_equations, x="complexity", y="train_r2", label="Training R2")
sns.scatterplot(symbolic_equations, x="complexity", y="val_r2", label="Validation R2")
plt.xlabel("Equation Complexity")
plt.ylabel("R2")
```

```python
symbolic_equations.sort_values("val_r2").tail(5)[["complexity", "equation"]].values
```

```python
regression_equation = '1.67113930389962*alphafold_iptm*boltz_chain_1_ptm*(boltz_chain_1_ptm + boltz_distance_score)*(boltz_peptide_plddt*(boltz_interface_sasa + hydrophobic_aa_percent) + intra_boltz_mean_rmsd) + 1.67113930389962*boltz_interface_sasa + 4.46809624612647'
```

```python
def symbolic_predict(df, scaling_parameters, equation='3.6232204*boltz_chain_1_ptm - boltz_interface_dG + 3.6232204*boltz_interface_sasa + 2.24350459347672'):
    """ Scale using parameters and min-max then predict using symbolic equation """
    df = df.copy()
    for _, row in scaling_parameters.iterrows():
        feature = row['Unnamed: 0']
        min_val = row['min']
        max_val = row['max']
        df[feature] = (df[feature] - min_val) / (max_val - min_val)
    symbolic_pred = eval(equation, {}, df.to_dict(orient='series'))
    return pd.DataFrame({'symbolic_pred': symbolic_pred, "complex_filename": df["complex_filename"]})

predictions = symbolic_predict(real_val, regression_scaling_parameters, regression_equation)
predictions = predictions.merge(real_val[["complex_filename", "pKd"]], on="complex_filename")
predictions.head()
```

```python

plt.figure(figsize=(4,4))
r2 = stats.pearsonr(predictions["symbolic_pred"], predictions["pKd"])[0]**2
print("Train R2:", r2)
sns.scatterplot(predictions, x="symbolic_pred", y="pKd")
sns.despine()
```

# Classification fit

```python
classification_scaling_parameters = pd.read_csv("/home/er8813ha/immunopeptides/plots_v2_both/classification/scaling_params.csv")
```

```python
symbolic_equations = pd.read_csv("/home/er8813ha/immunopeptides/plots_v2_both/classification/symbolic_classification_all_equations.csv")
symbolic_equations
```

```python
plt.figure(figsize=(4,4))
sns.scatterplot(symbolic_equations, x="complexity", y="train_auc", label="Training AUC")
sns.scatterplot(symbolic_equations, x="complexity", y="val_auc", label="Validation AUC")
plt.xlabel("Equation Complexity")
plt.ylabel("AUC")
```

```python
symbolic_equations.sort_values("val_auc").tail(10)[["complexity", "equation"]].values
```

```python
classification_equation = '0.65594006*alphafold_ipsae_max + 0.65594006*boltz_peptide_plddt - 0.125507393976584'
```

```python
predictions_real = symbolic_predict(real_val, classification_scaling_parameters, classification_equation)
predictions_random = symbolic_predict(random_x_val, classification_scaling_parameters, classification_equation)
predictions_shuffle = symbolic_predict(shuffle_x_val, classification_scaling_parameters, classification_equation)

predictions_random["type"] = "random"
predictions_shuffle["type"] = "shuffle"
predictions_real["type"] = "real"

all_predictions = pd.concat([predictions_real, predictions_random, predictions_shuffle])
```

```python
plt.figure(figsize=(4,4))
sns.kdeplot(all_predictions, x="symbolic_pred", hue="type", fill=True)
sns.despine()
```

# Test set combined

```python
real_x_test = pd.read_csv("/home/er8813ha/immunopeptides/data/x_y_v2/real_X_test.csv")
real_y_test = pd.read_csv("/home/er8813ha/immunopeptides/data/x_y_v2/real_y_test.csv")
shuffle_x_test = pd.read_csv("/home/er8813ha/immunopeptides/data/x_y_v2/shuffle_X_test.csv")
random_x_test = pd.read_csv("/home/er8813ha/immunopeptides/data/x_y_v2/random_X_test.csv")
real_test = real_x_test.merge(real_y_test, on="complex_filename")
```

```python
test_predictions_reg = symbolic_predict(real_test, regression_scaling_parameters, regression_equation)
test_predictions_reg = test_predictions_reg.merge(real_test[["complex_filename", "pKd"]], on="complex_filename")
plt.figure(figsize=(4,4))
r2 = stats.pearsonr(test_predictions_reg["symbolic_pred"], test_predictions_reg["pKd"])[0]**2
print("Test R2:", r2)
sns.regplot(test_predictions_reg, x="symbolic_pred", y="pKd")
sns.despine()
```

```python
test_predictions_classification = symbolic_predict(real_test, classification_scaling_parameters, classification_equation)
test_predictions_real = test_predictions_classification.merge(test_predictions_reg, on="complex_filename", suffixes=("_classif", "_reg"))
test_predictions_real["total_pred"] = test_predictions_real["symbolic_pred_classif"] * test_predictions_real["symbolic_pred_reg"]
plt.figure(figsize=(4,4))
r2 = stats.pearsonr(test_predictions_real["total_pred"], test_predictions_real["pKd"])[0]**2
print("Test R2:", r2)
sns.regplot(test_predictions_real, x="total_pred", y="pKd")
sns.despine()
```

```python
random_predictions_class = symbolic_predict(random_x_test, classification_scaling_parameters, classification_equation)
shuffle_predictions_class = symbolic_predict(shuffle_x_test, classification_scaling_parameters, classification_equation)
random_predictions_reg = symbolic_predict(random_x_test, regression_scaling_parameters, regression_equation)
shuffle_predictions_reg = symbolic_predict(shuffle_x_test, regression_scaling_parameters, regression_equation)

random_predictions = random_predictions_class.merge(
    random_predictions_reg,
    on="complex_filename",
    suffixes=("_classif", "_reg")
)
shuffle_predictions = shuffle_predictions_class.merge(
    shuffle_predictions_reg,
    on="complex_filename",
    suffixes=("_classif", "_reg")
)
shuffle_predictions["total_pred"] = shuffle_predictions["symbolic_pred_classif"] * shuffle_predictions["symbolic_pred_reg"]
random_predictions["total_pred"] = random_predictions["symbolic_pred_classif"] * random_predictions["symbolic_pred_reg"]

all_predictions = pd.concat([
    test_predictions_real.assign(type="real"),
    random_predictions.assign(type="random"),
    shuffle_predictions.assign(type="shuffle"),
])
all_predictions.head()
```

```python
plt.figure(figsize=(4,4))
sns.kdeplot(all_predictions, x="symbolic_pred_reg", hue="type", fill=True)
```
